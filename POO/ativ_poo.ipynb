```python
class Musica:
    playlist = []
    def __init__(self, nome, artista, duracao):
        self.nome = nome
        self.artista = artista 
        self.duracao = duracao
        Musica.playlist.append(self)

    def __str__(self):
        return f'{self.nome} | {self.artista} | {self.duracao}'
    
    @classmethod
    def listar_musicas(cls):
        for musica in cls.playlist:
            print(musica)

    @classmethod
    def exibir_total_playlist(cls):
        soma_total = 0
        for musica in cls.playlist:
            soma_total += musica.duracao
        print(f'A playlist tem {len(cls.playlist)} músicas e dura {soma_total:.2f} minutos')

Musica('Does Haven even know?', 'Nickelback', 3.43)
Musica('Photograph', 'Nickelback', 4.21)
Musica('Far Away', 'Nickelback', 3.58)

Musica.exibir_total_playlist()
Musica.listar_musicas()
```

```python
class EstoqueLaboratorio:
    def __init__(self, responsavel):
        self.__responsavel = responsavel
        self.__reagentes = {}

    def adicionar_reagente(self, nome: str, quantidade: int):
        if nome in self.__reagentes:
            self.__reagentes[nome] += quantidade
            print(f'Adicionado {quantidade} unidades de {nome} ao estoque.')
        else:
            self.__reagentes[nome] = quantidade

    def __verificar_disponibilidade(self, nome: str, quantidade: int) -> bool:
        return nome in self.__reagentes and self.__reagentes[nome] >= quantidade
    
    def consumir_reagente(self, nome: str, quantidade: int):
        if self.__verificar_disponibilidade(nome, quantidade):
            self.__reagentes[nome] -= quantidade
            print(f'Sucesso: {quantidade} unidades de {nome} consumidas.')
        else:
            print(f"Erro: Quantidade insuficiente de {nome} no estoque.")

estoque = EstoqueLaboratorio('Vicenius')

estoque.adicionar_reagente('Alcool', 10)
estoque.consumir_reagente('Alcool', 5)
```

### Exercício 3: Geometria e Validação - Preparação para Herança


Requisitos:

Crie a classe Triangulo com os atributos privados __lado_a, __lado_b e __lado_c.

Crie um método privado __validar_condicao_existencia(a, b, c). Para um triângulo existir, a soma de dois lados deve ser sempre maior que o terceiro lado. Se os parâmetros não formarem um triângulo, retorne False.

No construtor (init) e no método set_lados, utilize o validador privado. Se a validação falhar, os lados devem ser definidos como 0 e o programa deve avisar no console: "Valores inválidos. Triângulo não formado.".

Crie o método tipo() que retorna "Equilátero" (3 lados iguais), "Isósceles" (2 lados iguais) ou "Escaleno" (3 lados diferentes).

Crie um método de classe (@classmethod) chamado criar_equilatero(cls, tamanho) que recebe apenas um valor e instancia/retorna um triângulo com os três lados iguais àquele valor.

Teste Prático: Instancie um triângulo válido (ex: 3, 4, 5). Tente alterá-lo usando set_lados(1, 10, 2) e comprove que o Setter barrou a alteração através da regra de negócio. Use o método de classe para criar um triângulo de lado 6 e imprima o tipo() dele.

```python
class Triangulo:
    def __init__(self, a: float, b: float, c: float):
        self.set_lados(a, b, c)
            
    def __validar_condicao_existencia(self, a: float, b: float, c: float) -> bool:
        return (a + b > c) and (a + c > b) and (b + c > a)
    
    def get_lados(self) -> tuple:
        return self.__lado_a, self.__lado_b, self.__lado_c
    
    def set_lados(self, a, b, c):
        if self.__validar_condicao_existencia(a, b, c):
            self.__lado_a = a
            self.__lado_b = b
            self.__lado_c = c
        else:
            print("Erro: Valores inválidos. Triângulo não formado")
            self.__lado_a = 0
            self.__lado_b = 0
            self.__lado_c = 0

    def tipo(self) -> str:
        if self.__lado_a == self.__lado_b == self.__lado_c:
            return 'Equilátero'
        elif self.__lado_a == self.__lado_b or self.__lado_b == self.__lado_c or self.__lado_a == self.__lado_c:
            return 'Isósceles'
        else:
            return 'Escaleno'
        
    @classmethod
    def criar_equilatero(cls, tamanho):
        return cls(tamanho, tamanho, tamanho)
        
# 1. Instanciando um triângulo perfeitamente válido
print("--- Teste 1: Triângulo Válido ---")
meu_triangulo = Triangulo(3, 4, 5)
print(f"Lados: {meu_triangulo.get_lados()}")
print(f"Tipo: {meu_triangulo.tipo()}")

print("\n--- Teste 2: Triângulo Válido ---")
meu_triangulo2 = Triangulo(10, 10, 12)
print(f"Lados: {meu_triangulo2.get_lados()}")
print(f"Tipo: {meu_triangulo2.tipo()}")

# 2. Testando a barreira de segurança do set_lados
print("\n--- Teste 3: Tentando burlar a regra ---")
# Aqui a regra de negócio (soma de dois lados > terceiro) vai falhar
meu_triangulo.set_lados(1, 10, 2) 
print(f"Lados após a tentativa: {meu_triangulo.get_lados()}")

# 3 e 4. Testando o método de classe e imprimindo o tipo
print("\n--- Teste 4: A Fábrica de Equiláteros ---")
# Usando o @classmethod diretamente do nome da classe
triangulo_perfeito = Triangulo.criar_equilatero(7)
print(f"Lados do novo triângulo: {triangulo_perfeito.get_lados()}")
print(f"Tipo do novo triângulo: {triangulo_perfeito.tipo()}")
```

```python
class Microcontrolador:
    contador_instancias = 0
    __slots__ = ['fabricante', 'modelo', 'frequencia_mhz']

    def __init__(self, fabricante, modelo, frequencia_mhz):
        self.fabricante = fabricante
        self.modelo = modelo
        self.frequencia_mhz = frequencia_mhz
        Microcontrolador.contador_instancias += 1

    def get_detalhes(self):
      return self.fabricante, self.modelo, self.frequencia_mhz

    @classmethod
    def relatorio_total(cls):
      return cls.contador_instancias

    @staticmethod
    def converter_mhz_para_hz(mhz):
      return mhz * 1000000

# Testes
micro1 = Microcontrolador('Microchip', 'PIC18F4550', 48)
micro2 = Microcontrolador('Atmel', 'ATmega328P', 16)

print(f"Total de instâncias: {Microcontrolador.relatorio_total()}")
print(f"Frequência do micro1 em Hz: {Microcontrolador.converter_mhz_para_hz(micro1.frequencia_mhz)}")
```
